```python
# loading the required libraries
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
# loading the ipl matches data set
ipl=pd.read_csv('Ipldata.csv')
```

```python
# having a glance of the first five records of the data set
ipl.head()
```

```python
# looking at the no.of the rows and colomns in the data set
ipl.shape
```

```python
# getting the frequency of most man of the match awards
ipl['player_of_match'].value_counts()
```

```python
# getting the frequency of top 10 most man of the match awards
ipl['player_of_match'].value_counts()[0:10]
```

```python
# it gives keys only
list(ipl['player_of_match'].value_counts()[0:10].keys())
```

```python
# bar plot
plt.figure(figsize=(14,5))
plt.bar(list(ipl['player_of_match'].value_counts()[0:10].keys()),list(ipl['player_of_match'].value_counts()[0:10]),color='g')
plt.show()
```

```python
# getting the frequency of the result column
ipl['result'].value_counts()
```

```python
# getting the frequency of the winning the toss
ipl['toss_winner'].value_counts()
```

```python
# Extracting the records where a team won batting first
batting_first=ipl[ipl['win_by_runs']!=0]
batting_first
```

```python
# making histogram which team are choses the bat or ball after toss wining.
plt.figure(figsize=(5,5))
plt.hist(ipl['toss_decision'])
plt.title("Bat Vs Ball")
plt.xlabel("name")
plt.ylabel('Number')
plt.show()
```

```python
# winner teams
ipl['winner'].value_counts()
```

```python
# bar plot for top 3 teams with most wins
plt.figure(figsize=(8,7))
plt.bar(list(ipl['winner'].value_counts()[0:3].keys()),list(ipl['winner'].value_counts()[0:3]),color=["blue","yellow","red"])
plt.show()
```

```python
# same thing showing pie chart
plt.figure(figsize=(6,6))
plt.pie(list(ipl['winner'].value_counts()),labels=list(ipl['winner'].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```

```python
# batting second team won
batting_second = ipl[ipl['win_by_wickets']!= 0]
```

```python
batting_second
```

```python
# now creatong the histogram of frequency win w.r.t number of wickets
plt.figure(figsize=(5,5))
plt.hist(batting_second['win_by_wickets'],bins=30)
plt.xlabel("Wickets")
plt.ylabel("Matches")
plt.title("Teams and their wining ratios after second bat")
plt.show()
```

```python
# finding team who played bat second and won 
batting_second['winner'].value_counts()
```

```python
# making a bar plot
plt.figure(figsize=(5,5))
plt.bar((batting_second['winner'].value_counts()[0:3].keys()),list(batting_second['winner'].value_counts()[0:3]),color=["red","purple","blue"])
plt.show()
```

```python
# now making pie chart
plt.figure(figsize=(5,5))
plt.pie((batting_second['winner'].value_counts()),labels=list(batting_second['winner'].value_counts().keys()),autopct='%0.1f%%')
plt.show()
```
