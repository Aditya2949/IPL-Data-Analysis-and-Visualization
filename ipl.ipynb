```python
# loading the required libraries
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
# loading the ipl matches data set
ipl=pd.read_csv('Ipldata.csv')
```

```python
# having a glance of the first five records of the data set
ipl.head()
```

```python
# looking at the no.of the rows and colomns in the data set
ipl.shape
```

```python
# getting the frequency of most man of the match awards
ipl['player_of_match'].value_counts()
```

```python
# getting the frequency of top 10 most man of the match awards
ipl['player_of_match'].value_counts()[0:10]
```

```python
# it gives keys only
list(ipl['player_of_match'].value_counts()[0:10].keys())
```

```python
# bar plot
plt.figure(figsize=(14,5))
plt.bar(list(ipl['player_of_match'].value_counts()[0:10].keys()),list(ipl['player_of_match'].value_counts()[0:10]),color='g')
plt.show()
```

```python
# getting the frequency of the result column
ipl['result'].value_counts()
```

```python
# getting the frequency of the winning the toss
ipl['toss_winner'].value_counts()
```

```python
# Extracting the records where a team won batting first
batting_first=ipl[ipl['win_by_runs']!=0]
batting_first
```

```python
# making histogram which team are choses the bat or ball after toss wining.
plt.figure(figsize=(5,5))
plt.hist(ipl['toss_decision'])
plt.title("Bat Vs Ball")
plt.xlabel("name")
plt.ylabel('Number')
plt.show()
```

```python
# winner teams
ipl['winner'].value_counts()
```

```python
# bar plot for top 3 teams with most wins
plt.figure(figsize=(8,7))
plt.bar(list(ipl['winner'].value_counts()[0:3].keys()),list(ipl['winner'].value_counts()[0:3]),color=["blue","yellow","red"])
plt.show()
```
